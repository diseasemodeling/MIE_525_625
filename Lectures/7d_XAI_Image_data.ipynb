# Captum Attribution Methods — Detailed Summary

This notebook summarizes attribution methods in Captum, organized by the level at which they operate (Feature, Layer, Neuron).  
Each table lists the **method type**, **mathematical formulation**, **advantages**, and **limitations**, ordered **chronologically** with emphasis on methodological improvements.  

---

## 🧩 Feature Attribution Methods

| **Type** | **Description + Key Equation** | **Pros** | **Cons** |
|-----------|-------------------------------|-----------|-----------|
| **Saliency (Simonyan et al., 2013)**  <br>Local • Model-specific • Any data • Any DL model | Earliest gradient-based method. Computes gradient of model output \( F(x) \) w.r.t. input features.  <br><br>$$ S_i(x) = \frac{\partial F(x)}{\partial x_i} $$  <br>Provides first-order sensitivity to input changes. | Simple and efficient; provides intuition about sensitivity. | Noisy; sensitive to input scaling; does not handle nonlinearity or saturation. |
| **DeconvNet (Zeiler & Fergus, 2014)**  <br>Local • Model-specific • Image • CNN | Improves saliency for CNNs by modifying backward pass: passes only positive gradients through ReLU. <br><br>$$ \delta_{deconv} = \delta_{up} \cdot \mathbb{I}(a > 0) $$ | Produces sharper visualizations for CNN filters. | Not theoretically grounded; heuristic modification. |
| **Guided Backpropagation (Springenberg et al., 2014)**  <br>Local • Model-specific • Image • CNN | Further improves DeconvNet by blocking negative gradients **and** negative activations during ReLU backprop. <br><br>$$ \delta_{guided} = \delta_{up} \cdot \mathbb{I}(a>0)\mathbb{I}(\delta_{up}>0) $$ | High-resolution feature maps; easy to implement. | Still heuristic; not class-discriminative; fails for non-ReLU layers. |
| **Input × Gradient (Shrikumar et al., 2016)**  <br>Local • Model-specific • Any data • Any DL model | Multiplies input feature by its gradient; inspired by linear approximation of the model. <br><br>$$ \text{IxG}_i = x_i \cdot \frac{\partial F(x)}{\partial x_i} $$ | Simple, corrects for input scale; interpretable. | Fails for saturated regions (gradient ≈ 0). |
| **DeepLIFT (Shrikumar et al., 2017)**  <br>Local • Model-specific • Any data • Any DL model | Addresses gradient saturation by comparing activations to a reference baseline \(x'\). <br><br>$$ \text{Attribution}(x) = m_{x \to t} \Delta x, \quad m_{x \to t} = \frac{C_{\Delta x \to \Delta t}}{\Delta x} $$ | Handles zero gradients; satisfies completeness. | Requires reference choice; less stable for deep nonlinear networks. |
| **Integrated Gradients (Sundararajan et al., 2017)**  <br>Local • Model-specific • Any data • Any DL model | Integrates gradients along a straight path from baseline to input, ensuring axioms of sensitivity & implementation invariance. <br><br>$$ IG_i(x) = (x_i - x_i') \int_0^1 \frac{\partial F(x' + \alpha(x - x'))}{\partial x_i} \, d\alpha $$ | Theoretically principled; robust to scaling; baseline flexibility. | Computationally expensive (multiple gradient evaluations). |
| **Grad-CAM (Selvaraju et al., 2017)**  <br>Local • Model-specific • Image • CNN | Produces class-discriminative heatmaps by weighting feature maps with gradients. <br><br>$$ \alpha_k = \frac{1}{Z}\sum_{i,j}\frac{\partial y^c}{\partial A_{ij}^k}, \quad L^c = \text{ReLU}\left(\sum_k \alpha_k A^k\right) $$ | Localizes discriminative regions; interpretable for CNNs. | Low spatial resolution; not pixel-level attribution. |
| **Guided Grad-CAM (Selvaraju et al., 2017)**  <br>Local • Model-specific • Image • CNN | Combines high-res detail of Guided BP with class discrimination of Grad-CAM (elementwise product). | Detailed and class-specific visualization. | Complex pipeline; not easily extended to non-CNN models. |
| **Shapley Value Sampling (Lundberg & Lee, 2017)**  <br>Global • Model-agnostic • Any data • Any DL model | Theoretically sound attribution via Shapley values from cooperative game theory. <br><br>$$ \phi_i = \frac{1}{|\Pi|} \sum_{\pi \in \Pi} [F(x_{S_i^\pi \cup \{i\}}) - F(x_{S_i^\pi})] $$ | Fairness axioms; model-agnostic. | Exponential complexity; approximated by sampling. |
| **LIME (Ribeiro et al., 2016)**  <br>Local • Model-agnostic • Any data • Any DL model | Trains a simple local surrogate \( g \) near instance \( x \). <br><br>$$ \min_{g\in G} L(f,g,\pi_x) + \Omega(g) $$ | Intuitive local explanations; flexible. | Instability; sensitive to sampling and kernel width. |
| **KernelSHAP (Lundberg & Lee, 2017)**  <br>Local • Model-agnostic • Any data • Any DL model | Unifies LIME and SHAP; uses weighted linear regression to approximate Shapley values. <br><br>$$ \phi = \arg\min_{\phi}\sum_{z'}\pi_x(z')(f(h_x(z'))-\phi^Tz')^2 $$ | Combines local interpretability with fairness axioms. | Computationally heavy for many features. |
| **Feature Ablation (Zeiler & Fergus, 2014; Captum impl.)**  <br>Local • Model-agnostic • Any data • Any DL model | Replaces one feature/group with baseline and measures change in output. <br><br>$$ \text{FA}_i = F(x) - F(x_{\setminus i},x'_i) $$ | Conceptually simple; no gradients needed. | Slow for large inputs; baseline-dependent. |
| **Feature Permutation (Fisher et al., 2019)**  <br>Global • Model-agnostic • Any data • Any DL model | Permutes feature values and measures prediction change to assess global importance. | Model-agnostic; measures interaction effects. | Expensive for large datasets; correlation confounding. |
| **Occlusion (Zeiler & Fergus, 2014)**  <br>Local • Model-agnostic • Image • CNN | Masks regions and measures output change. <br><br>$$ \text{Occ}(p) = \frac{1}{N_p} \sum_{r: p \in r}[F(x) - F(x_{\text{occluded }r})] $$ | Intuitive visual interpretation. | Computationally expensive; window-size sensitive. |

## 🧠 Layer Attribution Methods

| **Type** | **Description + Equation** | **Pros** | **Cons** |
|-----------|----------------------------|-----------|-----------|
| **Layer Gradient (Simonyan, 2013)** | Gradient of output w.r.t. layer activations. | Simple; works as internal saliency. | Noisy; lacks completeness. |
| **Layer Input × Gradient (Shrikumar, 2016)** | Product of activation and its gradient in the chosen layer. | Efficient approximation of layer contribution. | Same issues as raw gradients. |
| **Layer Integrated Gradients (Sundararajan, 2017)** | Integrated Gradients computed for intermediate layer activations. <br><br>$$ LIG_j(x) = (a_j - a'_j)\int_0^1 \frac{\partial F(A' + \alpha(A - A'))}{\partial a_j} d\alpha $$ | Satisfies completeness; applicable to internal layers. | Requires baseline for activations. |
| **Layer Conductance (Dhamdhere et al., 2018)** | Measures contribution of a layer combining activations and gradients. <br><br>$$ \text{Cond}(A)=\sum_j (A_j-A'_j)\int_0^1 \frac{\partial F(A'+\alpha(A-A'))}{\partial A_j} d\alpha $$ | Quantifies how layer information flows to output. | Needs full activation path; baseline-dependent. |
| **Layer DeepLIFT (Shrikumar, 2017)** | Applies DeepLIFT propagation rules to intermediate layer activations. | Handles zero-gradient issues internally. | Same baseline challenges as DeepLIFT. |
| **Layer Gradient SHAP (Lundberg & Lee, 2017; Captum)** | SHAP-style expected gradients for intermediate layers. | Combines SHAP fairness with gradient efficiency. | Requires noise sampling; stochastic variance. |

## 🔬 Neuron Attribution Methods

| **Type** | **Description + Equation** | **Pros** | **Cons** |
|-----------|----------------------------|-----------|-----------|
| **Neuron Gradient (Simonyan, 2013)** | Gradient of neuron activation w.r.t. input. | Simple; helps understand neuron sensitivity. | Local linear approximation only. |
| **Neuron Input × Gradient (Shrikumar, 2016)** | Scales neuron’s input contribution by gradient. | Intuitive; reveals neuron influence. | Ignores inter-neuron dependencies. |
| **Neuron Integrated Gradients (Sundararajan, 2017)** | Integrates gradient for a specific neuron from baseline to input. <br><br>$$ NIG_i(x) = (x_i - x_i')\int_0^1 \frac{\partial a_n(x'+\alpha(x-x'))}{\partial x_i} d\alpha $$ | Satisfies completeness for neuron activations. | Requires baseline; expensive integration. |
| **Neuron Conductance (Dhamdhere et al., 2018)** | Decomposes layer conductance into neuron-level contributions. | Fine-grained neuron importance tracking. | Computationally heavy; depends on baseline. |
| **Neuron DeepLIFT (Shrikumar, 2017)** | Propagates relevance differences at neuron level. | Effective for explaining individual neuron activations. | Same baseline and saturation issues as DeepLIFT. |

## 🔁 Supplementary Sections

### **Noise Tunnel (Smilkov et al., 2017 — SmoothGrad Extension)**

Adds Gaussian noise to the input multiple times and averages the attributions to smooth out sharp gradients.

$$ \text{NT}(x) = \mathbb{E}_{\epsilon \sim \mathcal{N}(0,\sigma^2)} [ \text{Attribution}(x+\epsilon) ] $$

**Pros:** Reduces noise; stabilizes saliency maps.  
**Cons:** Computationally expensive (multiple forward/backward passes).

### **Attribution Metrics**

| **Metric** | **Equation** | **Interpretation** |
|-------------|--------------|--------------------|
| **Infidelity (Yeh et al., 2019)** | $$ \text{Infidelity} = \mathbb{E}_{I} \left[ \left( I^T \text{Attr}(x) - (F(x) - F(x - I)) \right)^2 \right] $$ | Measures how well attributions explain output changes under perturbations. Lower is better. |
| **Sensitivity (Ancona et al., 2018)** | $$ \text{Sensitivity} = \mathbb{E}_{\delta} [ \| \text{Attr}(x+\delta) - \text{Attr}(x) \| ] $$ | Quantifies attribution stability to small input noise. Lower indicates more robust attributions. |
