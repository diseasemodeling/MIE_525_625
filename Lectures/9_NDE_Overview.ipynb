# I. NDE Overview
* Time-series data occur in numerous applications. In many cases they are complex (high dimension state-space representation/ multiple interacting time-series signals) and non-linear. **CHaracteristics of real world data include irregular sampling, missing values, and continuous latent dynamics**.
* Traditional time-series models:
  * Inlcude: state space models (differntial equations, system dynamics), or auto-regressive (ARM, ARIMA, etc.)
  * Pros: Robust; Cons: assume regular sampling and often linear relationships
* Machine learning models:
  * Include: RNN, LSTM, Tranformers
  * Pros: capture nonlinearity; Cons: constrained by discrete-time formulations
* NDEs: offers principled framework for continuous-time modeling with neural networks
  * Since the seminal work by Chen et.al., 2018, on NODE, there have been significant advances, mostly focusing on benchmarked datasets/physical processes.
  * **Neural Ordinary Differential Equations (NODEs or NeuralODEs)** (Chen et al. 2018)
  * **Neural Controlled Differential Equations (NCDEs or NeuralCDEs)** (Kidger et al. 2020)
  * Neural Stochastic Differential Equations (NSDEs) (Han et al. 2017; Tzen and Raginsky 2019; Li et al. 2020; Oh et al. 2024).
  * Applications of these methods in real systems is more recent and rapdidly growing.








# II. NODE, NCDE, NSDE, UDE

**Review article for NODE --> NCDE --> NSDE**

* **NODE (Neural ODE)**  $$\frac{dz}{dt} = f_\theta(z, t)$$
  - Models continuous, deterministic evolution driven entirely by an initial condition $z(0)$.
  - Assumes **no prior knowledge (Black-box)**, so the neural network learns the entire vector field from scratch.
* **NCDE (Neural CDE / NCE)**  $$dz_t = f_\theta(z_t) \, dX_t$$
  * Extends NODEs so the trajectory continuously updates in response to an incoming, irregular data stream (control path $X_t$).
  * Assumes **no prior knowledge (Black-box)** and is designed for irregular time series where new observations steer the state.
* **NSDE (Neural SDE)**  $$dz_t = f_\theta(z_t, t)\,dt + g_\theta(z_t, t)\,dW_t$$
  * Adds a diffusion term ($g_\theta \, dW_t$) driven by Brownian motion to model intrinsic randomness and uncertainty.
  * Assumes **no prior knowledge (Black-box)**, learning both deterministic drift and stochastic volatility from data.
* **UDE for Scientific Machine Learning (UDE for SciML)**  $$\frac{dz}{dt} = f_{\text{known}}(z, t, U_\theta(z, t))$$
  * A **hybrid ("gray-box") framework** that keeps known scientific equations intact and uses a neural network ($U_\theta$) only for the unknown missing terms.
  * Assumes **partial prior knowledge (Gray-box)**, bridging pure physics models and pure neural differential equations.
  * Rackauckas, C., et al., 2020 https://arxiv.org/abs/2001.04385
